```python
import os

def list_folders(relative_path):
    """Lists all folders in the specified directory given by a relative path.
    
    Args:
        relative_path (str): The relative path to the directory.
        
    Returns:
        list: A list of folder names in the specified directory.
    """
    
    # Get the absolute path from the relative path
    absolute_path = os.path.abspath(relative_path)
    
    # List all folders in the directory
    try:
        folder_list = [
            name for name in os.listdir(absolute_path)
            if os.path.isdir(os.path.join(absolute_path, name))
        ]
        return folder_list
    except FileNotFoundError:
        print(f"Directory '{relative_path}' does not exist.")
        return []
    except Exception as e:
        print(f"An error occurred: {e}")
        return []

# Example usage:
# Suppose you want to list folders in the 'example_directory' relative to the script location
# folders = list_folders('example_directory')
# print(folders)
```

```python
import pandas as pd

def generate_intermediate_dataframes(experiment_number, consumer_id, base_dir=BASE_DIR):
    df_output_step = pd.read_json(OUTPUT_STEP.format(base_dir=base_dir ,consumer_id=consumer_id, experiment_number=experiment_number))
    df_search_step = pd.read_json(SEARCH_STEP.format(base_dir=base_dir, consumer_id=consumer_id, experiment_number=experiment_number))
    df_token_step = pd.read_json(TOKEN_STEP.format(base_dir=base_dir, consumer_id=consumer_id, experiment_number=experiment_number))
    df_tool_step = pd.read_json(TOOL_STEP.format(base_dir=base_dir, consumer_id=consumer_id, experiment_number=experiment_number))

    df_1 = df_output_step.merge(df_search_step, on='experiment_id', suffixes=('_output_step', '_search_step'))
    df_2 = df_token_step.merge(df_tool_step, on='experiment_id', suffixes=('_token_step', '_tool_step'))
    df_final = df_1.merge(df_2, on='experiment_id')  
    return df_final
```

```python
import pandas as pd
import os

BASE_DIR = "/home/baptvit/Documents/mestrado/master-experiments/logs"
EVAL_DIR = "/home/baptvit/Documents/mestrado/master-experiments/evaluations/data/silver"

OUTPUT_STEP = '{base_dir}/{consumer_id}/{model_id}/{experiment_id}/{question_id}/output_step.json'
SEARCH_STEP = '{base_dir}/{consumer_id}/{model_id}/{experiment_id}/{question_id}/search_step.json'
TOKEN_STEP = '{base_dir}/{consumer_id}/{model_id}/{experiment_id}/{question_id}/tokens_step.json'
TOOL_STEP = '{base_dir}/{consumer_id}/{model_id}/{experiment_id}/{question_id}/tool_step.json'

SAVE_INTERMEDIATE_DF = os.path.join(EVAL_DIR, "{consumer_id}.csv")
```

```python
def generate_intermediate_dataframes(list_consumers_id, list_models, list_experiments, list_questions, base_dir=BASE_DIR):
    all_dfs = []
    for consumer_id in list_consumers_id:
        for model_id in list_models:
            for experiment_id in list_experiments:
                for question_id in list_questions:
                    if experiment_id in question_id and model_id in question_id:
                        try:
                            output_path = OUTPUT_STEP.format(base_dir=base_dir, consumer_id=consumer_id, model_id=model_id, experiment_id=experiment_id, question_id=question_id)
                            search_path = SEARCH_STEP.format(base_dir=base_dir, consumer_id=consumer_id, model_id=model_id, experiment_id=experiment_id, question_id=question_id)
                            token_path = TOKEN_STEP.format(base_dir=base_dir, consumer_id=consumer_id, model_id=model_id, experiment_id=experiment_id, question_id=question_id)
                            tool_path = TOOL_STEP.format(base_dir=base_dir, consumer_id=consumer_id, model_id=model_id, experiment_id=experiment_id, question_id=question_id)
    
                            df_output_step = pd.read_json(output_path)
                            df_search_step = pd.read_json(search_path)
                            df_token_step = pd.read_json(token_path)
                            df_tool_step = pd.read_json(tool_path)
    
                            df_1 = df_output_step.merge(df_search_step, on='experiment_id', suffixes=('_output_step', '_search_step'))
                            df_2 = df_token_step.merge(df_tool_step, on='experiment_id', suffixes=('_token_step', '_tool_step'))
                            df_final = df_1.merge(df_2, on='experiment_id')
                            all_dfs.append(df_final)
    
                        except FileNotFoundError:
                            print(f"Files not found for: {output_path}, {search_path}, {token_path} e {tool_path}")
                            continue # Skip to the next combination if files are missing
                        except pd.errors.EmptyDataError:
                            print(f"Empty JSON files for: Consumer: {consumer_id}, Model: {model_id}, Experiment: {experiment_id}, Question: {question_id}")
                            continue
                        except KeyError as e:
                            print(f"KeyError: {e} in files for: Consumer: {consumer_id}, Model: {model_id}, Experiment: {experiment_id}, Question: {question_id}. Check if 'experiment_id' exists in all dataframes.")
                            continue
                        except Exception as e:
                            print(e)
    if all_dfs:
        final_df = pd.concat(all_dfs, ignore_index=True)
        return final_df
    else:
      return pd.DataFrame() # Return empty dataframe if no data was processed
```
