## Count of resources by consumer ID

```python
MATCH (n)
RETURN 
  n.consumer_id AS consumer_id,
  n.resource_type AS resource_type,
  COUNT(*) AS count
ORDER BY 
  n.consumer_id ASC, 
  count DESC
```

```python
import pandas as pd
```

```python
df = pd.read_csv("./count_resources_by_consumer_id_v1.csv")
```

```python
df
```

```python
list_resources = ["Condition", "Observation", "Encounter", "MedicationRequest", "Procedure", "DiagnosticReport", "Claim", "ExplanationOfBenefit", "Immunization", "AllergyIntolerance", "CarePlan"] 
```

```python
target_df = df[df["resource_type"].isin(list_resources)].sort_values(["consumer_id", "resource_type"])
```

```python
pivoted = target_df.pivot(index="consumer_id", columns="resource_type", values="count")
pivoted
```
