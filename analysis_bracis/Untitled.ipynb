```python
import json

with open(path) as f:
    d = json.load(f)
```

```python
list_resources = []
for entry in range(len(d["entry"])):
    resource = d["entry"][entry]["resource"]
    list_resources.append({"resource": str(resource)})
```

```python
import json
with open('./output.json', 'w') as fout:
    json.dump(list_resources, fout)
```
